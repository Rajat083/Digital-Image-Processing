```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
np.__version__
```

```python
img = cv2.imread('Images/lambo.jpg')
type(img)
```

```python
def print_img(img, note=""):
    plt.title(f"Lamborghini Urus {note}: res: {img.shape[0]}x{img.shape[1]}")
    plt.imshow(img)
```

```python
print_img(img)
```

```python
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
print_img(img, "grayscale")
```

```python
def sampling(img: np.ndarray, factor: int=2) -> np.ndarray: 
    return img[::factor, ::factor]
```

```python
sampled_img = sampling(img, 4)
print_img(sampled_img, "Sampled factor 4")
```

```python
sampled_img = sampling(img, 2)
print_img(sampled_img, "Sampled factor 4")
```

```python
sampled_img = sampling(img, 8)
print_img(sampled_img, "Sampled factor 4")
```

```python
def quantization(img, levels=4):
    img_max = img.max()
    quantized_img = img[:, :]
    quantized_img = ((quantized_img / img_max )* levels).astype(int)
    # for i, row in enumerate(quantized_img):
    #     for j, _ in enumerate(row):
    #         quantized_img[i, j] = int(quantized_img[i, j] / img_max) * levels 
    # quantized_img = int(img / img_max) * levels
    return quantized_img
```

```python
quant = quantization(img, 16)
print_img(quant, "Quantized")
```
