# Lab03 Image Transformations

```python
import numpy as np 
import cv2 
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("Images/porsche.jpg", cv2.IMREAD_GRAYSCALE)
img.shape
```

```python
def print_img(img, note=""):
    plt.title(f"Porsche Turbo S {note}: res: {img.shape[0]}x{img.shape[1]}")
    plt.imshow(img)
```

```python
print_img(img)
```

```python
def img_negative_transform(img):
    L = img.max()
    return (L - img).astype(int)
```

```python
neg_img = img_negative_transform(img)
print_img(neg_img, "Negative Transform")
```

```python
def contrast(img):
    exp = 1 + np.exp(-img)
    sigmoid = np.power(exp, -1)
    result = (sigmoid * 100).astype(int)
    return result
```

```python
img_contrast = contrast(img)
print_img(img_contrast, "Contrasted")
```

```python
def log_transform(img, c=10):
    img = img.astype(float)
    return (c * (np.log(1 + img) / np.log(10))).astype(int)
```

```python
log_img = log_transform(img, 10)
print_img(log_img, "Log Transformed")
```

```python
def power_transform(img, a, g):
    return a * np.pow(img, g).astype(int)
```

```python
pow_img= power_transform(img, 10, 0.3)
print_img(pow_img, "Power transformed")
```
