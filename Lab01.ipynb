```python
import cv2
import os

print(cv2.__version__)
```

```python
IMG_DIR="Images/"
```

```python
img = cv2.imread(os.path.join(IMG_DIR, "BMW.jpeg"))
cv2.imshow("Output Window", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imshow("GRAY SCALE", img_gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img_resized = cv2.resize(img_gray, (800, 600), interpolation=cv2.INTER_LINEAR)
cv2.imshow("GRAY SCALE RESIZED", img_resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
